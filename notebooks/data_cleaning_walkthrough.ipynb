# Data Cleaning & Exploration Walkthrough

This notebook demonstrates how to load, inspect, clean, and export tabular data using the modular `src.cleaner` and `src.utils` tools.

```python
import sys
sys.path.append('..')

from src.utils import load_dataset, inspect_data, print_summary
from src.cleaner import DataCleaner
```

## 1. Load Raw Dataset

```python
df = load_dataset('../data/raw/sample.csv')
df
```

## 2. Inspect Data Quality

```python
summary = inspect_data(df)
print_summary(summary)
```

## 3. Clean the Data

```python
cleaner = (
    DataCleaner(df)
    .clean_column_names()
    .trim_whitespace()
    .remove_duplicates()
    .handle_missing_values(strategy='drop')
)

cleaned_df = cleaner.get_data()
cleaned_df
```

## 4. Export Cleaned Dataset

```python
cleaner.export('../data/processed/cleaned_sample.csv')
```
